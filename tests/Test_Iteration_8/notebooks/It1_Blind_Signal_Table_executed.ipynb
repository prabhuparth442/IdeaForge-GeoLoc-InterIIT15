300 fresh NUDT photos (pool 2400; the 500-photo pool and every 404 / 505 / 606 photo excluded), dumped by `run_blind_dump.sh` with the T6
`Rich_Dump_Run` notebook. This notebook only re-fits the stored matches into the per-candidate signal table (block
`train/checks/decision_signals.py`, as in T7 It1). **No score is computed here**: the table is read only after `results/preregistered_t8.json`
is written (It5).

```python
import os, sys, time
from pathlib import Path
ROOT = Path.cwd()
while not (ROOT / "single_photo_pipeline.py").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT)); sys.path.insert(0, str(ROOT / "src")); os.chdir(ROOT)
import pandas as pd
from train.checks.decision_signals import build_signal_table
DUMPS = ROOT / "tests/Test_Iteration_6/results/rich_dump"
OUT = ROOT / "tests/Test_Iteration_8/results/signals"
OUT.mkdir(parents=True, exist_ok=True)
```

```python
t = time.time()
table = build_signal_table(DUMPS / "nudt_fresh707", "nudt707", int(os.environ.get("WORKERS", "12")))
table.to_csv(OUT / "nudt707.csv.gz", index=False)
dict(photos=table.photo.nunique(), candidates=len(table), seconds=round(time.time() - t, 1), refit_equals_stored=round((table.inliers == table.inl_s3).mean(), 4))
```
