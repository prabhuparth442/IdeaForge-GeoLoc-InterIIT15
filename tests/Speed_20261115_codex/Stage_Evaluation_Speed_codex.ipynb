# Speed experiment: stage evaluation

CPU measurements; original and changed code use frozen weights and the same queries. Run cells to refresh progress. See REPORT_codex.md for limitations.


```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd()
while not (root / "single_photo_pipeline.py").exists():
    if root == root.parent: raise RuntimeError("Open this notebook within the IdeaForge repository")
    root = root.parent
experiment = root / "tests/Speed_20261115_codex"
results = experiment / "results"
```

```python
summary = pd.read_json(results / "summary_codex.json")
display(summary.drop(columns=["stage_mean"], errors="ignore"))
stages = pd.DataFrame([dict(domain=r["domain"], variant=r["variant"], **r["stage_mean"]) for r in summary.to_dict("records")])
display(stages)
```

```python
if len(stages):
    axes = stages.set_index(["domain", "variant"]).plot.bar(stacked=True, figsize=(10,4), ylabel="CPU seconds / photo", title="Measured stages (same queries and candidates)")
```

```python
historical = pd.read_csv(root / "tests/Stage_Evaluation/time_per_query_official_edge_zncc_headings9_scales3_K50_M10.csv")
steady = historical[historical.photo != historical.photo.min()]
display(steady[["render", "maps", "coarse", "semantic", "fine", "total"]].mean().to_frame("Historical CUDA seconds/query — different hardware/configuration context"))
```
